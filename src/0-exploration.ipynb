# Exploração de dados

```python
from pprint import pprint

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import config
```

```python
bunch = config.load_mnist()

data_df: pd.DataFrame = bunch.data
target_df: pd.Series = bunch.target

print(f'data_df: {type(data_df)}{data_df.shape}\ntarget_df: {type(target_df)}{target_df.shape}')
```

```python
pprint(bunch.details)
pprint(bunch.DESCR)
```

```python
samples = data_df.sample(8)
fig, ax = plt.subplots(2, 4, figsize=(8, 6))

for i, img in enumerate(samples.values):
    row = 0 if i < 4 else 1
    col = i % 4
    img_arr = img.reshape((28, 28))
    ax[row, col].imshow(img_arr, cmap=plt.cm.Grays, vmin=0, vmax=255, interpolation='none')

plt.show()
```

```python
mean = data_df.mean().values.reshape((28, 28))
plt.imshow(mean, cmap=plt.cm.Grays, vmin=0, vmax=255, interpolation='none')
```
